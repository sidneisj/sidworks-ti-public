## 1. Clonar o repositório e instalar as dependências

O clone é público e feito em `/content/sidworks-ti-public`. Não precisa montar seu Google Drive.


```python
from pathlib import Path
import subprocess
import sys

REPO = Path('/content/sidworks-ti-public')
URL = 'https://github.com/sidneisj/sidworks-ti-public.git'

if not REPO.exists():
    subprocess.run(['git', 'clone', '--depth', '1', URL, str(REPO)], check=True)
elif (REPO / '.git').is_dir():
    subprocess.run(['git', '-C', str(REPO), 'pull', '--ff-only'], check=True)
else:
    raise RuntimeError(f'O caminho {REPO} existe, mas não é um repositório Git.')

subprocess.run([sys.executable, '-m', 'pip', 'install', '-r', str(REPO / 'requirements.txt')], check=True)
print('Dependências instaladas.')
```

## 2. Executar a demonstração

A aplicação processa vagas fictícias de `examples/`, classifica os registros, remove duplicatas e produz dois documentos. Nenhuma API externa é chamada pelo processamento da demonstração.


```python
subprocess.run([sys.executable, 'main.py', '--demo'], cwd=REPO, check=True)

pdf = REPO / 'exports' / 'SidWorksTI_Demonstracao.pdf'
docx = REPO / 'exports' / 'SidWorksTI_Demonstracao_WhatsApp.docx'
assert pdf.is_file() and docx.is_file(), 'Relatórios não encontrados.'
print('Relatórios disponíveis:', pdf.name, 'e', docx.name)
```

## 3. Baixar os relatórios (opcional)

Execute a célula a seguir somente quando quiser baixar os arquivos PDF e Word para seu computador. Também é possível acessar a pasta `sidworks-ti-public/exports/` no painel de arquivos do Colab.


```python
from google.colab import files
files.download(str(pdf))
files.download(str(docx))
```
